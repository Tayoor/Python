# **Hexagonal Binning**

```python
df.plot.scatter(x='Duration (hours)', y='Price ($)')
plt.xlabel('Duration (hours)')
plt.ylabel('Price ($)')
plt.title('Scatter Plot of Duration (hours) vs Price ($)')
```

```python
df.plot.hexbin(x='Duration (hours)', y='Price ($)', gridsize=50)
plt.xlabel('Duration (hours)')
plt.ylabel('Price ($)')
plt.title('Hexbin Plot of Duration (hours) vs Price ($)')
```

### **Scatterplot vs Hexbin**
If we are ploting small data of two or more numeric variables then **scatterplot** is good for it and when we are ploting biggest size of numeric variables then **Hexbin** is better for it. if we use a bigdata on **scatterplot** will be too dense, so we need a
different way to visualize the relationship.

#**Contours**

```python
sns.kdeplot(x=df['Duration (hours)'],y=df['Price ($)'], cmap="inferno", fill=True, levels=10, thresh=0)

plt.xlabel('Duration (hours)')
plt.ylabel('Price ($)')
plt.title('Contour Plot of Duration (hours) vs Price ($)')

plt.show()
```

#**Violin Plots**

```python
sns.violinplot(x=df['Platform'],y=df['Price ($)'], inner='quartile', color='white')
plt.xlabel('Platform')
plt.ylabel('Price ($)')
plt.title('Violin Plot of Platform vs Price ($)')

plt.show()
```
