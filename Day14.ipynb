```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Age': [42, 23, 22, 47, 50, 60],
    'Glucose': [98, 68, 73, 79, 88, 82]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
correlation = df.corr()
correlation
```

```python
plt.scatter(df['Age'],df['Glucose'])
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.title('Correlation of Age and Glucose')
```
